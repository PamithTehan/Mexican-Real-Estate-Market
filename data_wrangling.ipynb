## Data Wrangling (L1)

### Ingesting multi-file listings, handling missing values, standardizing currencies into USD, and structuring tidy datasets.


```python
#Import pandas and numpy libraries

import pandas as pd
import numpy as np 
```

### Load first dataset from CSV file

```python
#Load the 1st dataset

firstPath = "./data/raw/mexico-real-estate-1.csv"

firstDf = pd.read_csv(firstPath)

firstDf.head()
```

```python
#information about the first dataset
firstDf.info()

firstDf.shape
```

```python
#Number of rows with NaN values in the first dataset
nanRowCount = firstDf.isna().sum()
print(nanRowCount)
```

### Cleaning NaN values in lat and lon columns

```python
firstDf = firstDf.dropna()

firstDf.head()
```

```python
#Updated information about the first dataset after dropping NaN values
firstDf.info()

firstDf.shape
```

### price_usd column transformation

```python
firstDf = firstDf.assign(
    price_usd = lambda x : (
                    x["price_usd"]
                    .str.replace("$", "", regex=False) #regex=False is used to avoid warning message
                    .str.replace(",", "", regex=False) 
                    .astype(float)
                    )
    )

firstDf.head()
```

```python
#final information about the first dataset after dropping NaN values and converting price_usd to float
firstDf.info()

firstDf.shape
```

### Save Cleaned first dataset as a CSV file


```python
firstDf.to_csv("./data/processed/mexico-real-estate-1-cleaned.csv", index=False)
```
