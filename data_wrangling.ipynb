## Data Wrangling (L1)

### Ingesting multi-file listings, handling missing values, standardizing currencies into USD, and structuring tidy datasets.


```python
#Import pandas libraries

import pandas as pd
```

### Load first dataset from CSV file

```python
firstPath = "./data/raw/mexico-real-estate-1.csv"

firstDf = pd.read_csv(firstPath)

firstDf.head()
```

```python
#information about the first dataset
firstDf.info()

firstDf.shape
```

```python
#Number of rows with NaN values in the first dataset
nanRowCount = firstDf.isna().sum()
print(nanRowCount)
```

### Cleaning Missing values of lat and lon columns in first Dataset

```python
firstDf = firstDf.dropna()

firstDf.head()
```

```python
#Updated information about the first dataset after dropping NaN values
firstDf.info()

firstDf.shape
```

### price_usd column transformation

```python
firstDf = firstDf.assign(
    price_usd = lambda x : (
                    x["price_usd"]
                    .str.replace("$", "", regex=False) #regex=False is used to avoid warning message
                    .str.replace(",", "", regex=False) 
                    .astype(float)
                    )
    )

firstDf.head()
```

```python
#final information about the first dataset after dropping NaN values and converting price_usd to float
firstDf.info()

firstDf.shape
```

### Save Cleaned first dataset as a CSV file


```python
firstDf.to_csv("./data/processed/mexico-real-estate-1-cleaned.csv", index=False)
```

### Load second dataset from CSV file

```python
secondPath = "./data/raw/mexico-real-estate-2.csv"

secondDf = pd.read_csv(secondPath)

secondDf.head()
```

```python
#information about the second dataset
secondDf.info()

print("\nMissing values in the second dataset \n")
print( secondDf.isna().sum() )

secondDf.shape
```

### Clean Missing Values of lan and lot columns in second Dataset

```python
#drop rows with NaN values in the second dataset
secondDf = secondDf.dropna()

secondDf.head()
```

### Convert price column from mxn to USD

```python
#create a new column price_usd in the second dataset by converting price_mxn to USD
exchange_rate = 18.5

secondDf = secondDf.assign(
    price_usd = lambda x: (
        x['price_mxn'].div(exchange_rate).round(4)) 
    )

secondDf = secondDf.drop(columns=['price_mxn'])

secondDf.head()
```

```python
#final information about the second dataset after dropping NaN values and converting price_mxn to price_usd
secondDf.info()

secondDf.shape
```

### Save Cleaned second dataset as a CSV file

```python
secondDf.to_csv("./data/processed/mexico-real-estate-2-cleaned.csv", index=False)
```
